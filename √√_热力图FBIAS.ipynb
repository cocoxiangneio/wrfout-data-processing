```python
import os
import pandas as pd
import numpy as np

# 获取文件夹中所有 Excel 文件的列表
folder_path = 'C:/Users/xmxm/Desktop/汇总制图'
excel_files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.endswith('.xlsx')]

# 初始化存储数据的字典
data_dict = {
    'odd': [],
    'even': []
}

# 读取每个 Excel 文件中的指定数据范围
for file in excel_files:
    df = pd.read_excel(file, sheet_name=0)
    subset = df.iloc[14:19, 11:19]
    
    # 提取方案号
    scheme = os.path.basename(file).split('_')[1].split('.')[0]
    
    # 判断方案号是单数还是双数
    if int(scheme[1:]) % 2 == 0:
        data_dict['even'].append(subset)
    else:
        data_dict['odd'].append(subset)

# 计算每个单元格的平均值
def calculate_average(data_list):
    # 将列表中的数据帧堆叠起来
    stacked_data = np.stack([df.values for df in data_list], axis=0)
    
    # 计算平均值，忽略 NaN 值
    return np.nanmean(stacked_data, axis=0)

odd_average = calculate_average(data_dict['odd'])
even_average = calculate_average(data_dict['even'])

# 将结果转换为 DataFrame 便于展示
odd_average_df = pd.DataFrame(odd_average, columns=[f'Unnamed: {i}' for i in range(11, 19)])
even_average_df = pd.DataFrame(even_average, columns=[f'Unnamed: {i}' for i in range(11, 19)])

print("Odd scheme average:")
print(odd_average_df)

print("\nEven scheme average:")
print(even_average_df)
```
