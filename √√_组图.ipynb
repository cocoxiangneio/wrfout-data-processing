```python
from PIL import Image
import os
import glob
from natsort import natsorted  # 导入 natsorted 函数

# 设定存放图片的文件夹路径
folder_path = "C:/Users/xmxm/Desktop/降水图"

# 获取文件夹中所有图片的文件名，并进行自然排序
image_files = natsorted(glob.glob(os.path.join(folder_path, '*.png')))

# 设定每行图片的数量和输出大图的尺寸
images_per_row = 4
output_width = 800  # 输出大图的宽度，可以根据需要调整
margin = 10  # 图片之间的间隔

# 计算每张小图的宽度和高度
image_width = (output_width - (images_per_row - 1) * margin) // images_per_row
image_height = image_width  # 假设每张小图都是正方形

# 创建一个新的大图
num_images = len(image_files)
num_rows = (num_images + images_per_row - 1) // images_per_row  # 向上取整得到行数
output_height = num_rows * (image_height + margin) - margin  # 计算大图的高度

# 创建一个白色背景的大图
result_image = Image.new('RGB', (output_width, output_height), (255, 255, 255))

# 将每张小图粘贴到大图中的正确位置
x_offset = 0
y_offset = 0
for i, image_file in enumerate(image_files):
    if i % images_per_row == 0 and i != 0:
        y_offset += image_height + margin
        x_offset = 0
    # 打开并调整大小
    img = Image.open(image_file)
    img = img.resize((image_width, image_height), Image.LANCZOS)  # 调整大小并进行平滑处理
    # 粘贴到大图中
    result_image.paste(img, (x_offset, y_offset))
    x_offset += image_width + margin

# 保存生成的大图
result_image.save("C:/Users/xmxm/Desktop/降水图/output_image.png")
```
