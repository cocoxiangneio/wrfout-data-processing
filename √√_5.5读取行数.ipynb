```python
###读取一个文件夹列数
import pandas as pd

# Load the Excel file
file_path = "E:/000_FuJian/00_检验_run/03NA_0.1.xlsx"
excel_data = pd.ExcelFile(file_path)

# Initialize a list to store the results
data = []

# Iterate through each sheet in the workbook
for sheet_name in excel_data.sheet_names:
    # Read the sheet
    sheet_data = pd.read_excel(excel_data, sheet_name=sheet_name, header=None)
    
    # Extract the content of the first row and the fourth column
    fourth_column_first_row = sheet_data.iloc[0, 3]

    # Get the number of columns
    num_columns = sheet_data.shape[1]

    # Append the results to the data list
    data.append([fourth_column_first_row, num_columns])

# Create a DataFrame from the data
df = pd.DataFrame(data, columns=["Fourth column first row value", "Number of columns"])

# Save the DataFrame to a new Excel file
output_file_path = "E:/000_FuJian/00_检验_run/output.xlsx"
df.to_excel(output_file_path, index=False)

print("Output saved to", output_file_path)
```

```python
import os
import pandas as pd

# Define the directory containing the Excel files
directory_path = "E:/000_FuJian/00_检验_run"

# Get a list of all Excel files in the directory
excel_files = [f for f in os.listdir(directory_path) if f.endswith('.xlsx')]

# Iterate through each Excel file in the directory
for excel_file in excel_files:
    # Construct the full file path
    file_path = os.path.join(directory_path, excel_file)
    
    # Load the Excel file
    excel_data = pd.ExcelFile(file_path)

    # Initialize a list to store the results
    data = []

    # Iterate through each sheet in the workbook
    for sheet_name in excel_data.sheet_names:
        # Read the sheet
        sheet_data = pd.read_excel(excel_data, sheet_name=sheet_name, header=None)
        
        # Extract the content of the first row and the fourth column if it exists
        try:
            fourth_column_first_row = sheet_data.iloc[0, 3]
        except IndexError:
            fourth_column_first_row = None

        # Get the number of rows
        num_rows = sheet_data.shape[0] -1

        # Append the results to the data list
        data.append([fourth_column_first_row, num_rows])

    # Create a DataFrame from the data
    df = pd.DataFrame(data, columns=["Fourth column first row value", "Number of rows"])

    # Construct the output file path
    output_file_name = os.path.splitext(excel_file)[0] + '-1.xlsx'
    output_file_path = os.path.join(directory_path, output_file_name)
    
    # Save the DataFrame to a new Excel file
    df.to_excel(output_file_path, index=False)

    print(f"Output saved to {output_file_path}")
```
