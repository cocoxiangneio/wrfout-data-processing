```python
import numpy as np  
import matplotlib.pyplot as plt  
  
# 修正后的RGB值列表  
rgb_replaced = [  
    [255, 255, 255], [165, 221, 236], [113, 177, 226], [133, 196, 63],  
    [110, 189, 68], [246, 235, 0], [253, 184, 26], [246, 139, 30],  
    [237, 31, 35], [164, 40, 41], [167, 68, 152]  
]  
  
# 将RGB值转换为0到1之间的浮点数  
colors = np.array(rgb_replaced) / 255.0  
  
# 降雨量级别  
clevs = [0.1, 1, 2, 5, 7.5, 10, 13, 16, 20, 25, 30, 35, 40, 50, 60,  
         70, 80, 90, 100, 125, 150, 175, 200, 250]  
  
# 创建figure和axes对象  
fig, ax = plt.subplots()  
  
# 创建一个Normalized对象，用于将数据映射到颜色空间  
norm = plt.Normalize(vmin=min(clevs), vmax=max(clevs))  
  
# 创建一个ScalarMappable对象，用于生成颜色条  
cmap = plt.cm.colors.ListedColormap(colors)  
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)  
  
# 绘制颜色条  
cbar = fig.colorbar(sm, ax=ax, ticks=clevs, orientation='horizontal',  
                    spacing='proportional', drawedges=True)  
  
# 设置颜色条的标签  
cbar.set_label('Precipitation (mm)')  
  
# 显示图表  
plt.show()
```

```python
import numpy as np  
import matplotlib.pyplot as plt  
  
# 修正后的RGB值列表  
rgb_replaced = [  
    [255, 255, 255], [165, 221, 236], [113, 177, 226], [133, 196, 63],  
    [110, 189, 68], [246, 235, 0], [253, 184, 26], [246, 139, 30],  
    [237, 31, 35], [164, 40, 41], [167, 68, 152]  
]  
  
# 将RGB值转换为0到1之间的浮点数  
colors = np.array(rgb_replaced) / 255.0  
  
# 降雨量级别  
clevs = [0.1, 1, 2, 5, 7.5, 10, 13, 16, 20, 25, 30, 35, 40, 50, 60,  
         70, 80, 90, 100, 125, 150, 175, 200, 250]  
  
# 选择要在颜色条上显示的数值  
ticks_to_show = [0.1, 10,  50, 100, 200, 250]  
  
# 创建figure和axes对象  
fig, ax = plt.subplots()  
  
# 创建一个Normalized对象，用于将数据映射到颜色空间  
norm = plt.Normalize(vmin=min(clevs), vmax=max(clevs))  
  
# 创建一个ScalarMappable对象，用于生成颜色条  
cmap = plt.cm.colors.ListedColormap(colors)  
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)  
  
# 绘制颜色条，并指定要显示的数值  
cbar = fig.colorbar(sm, ax=ax, ticks=ticks_to_show, orientation='horizontal',  
                    spacing='proportional', drawedges=True)  
  
# 设置颜色条的标签  
cbar.set_label('Precipitation (mm)')  
  
# 显示图表  
plt.show()
```
