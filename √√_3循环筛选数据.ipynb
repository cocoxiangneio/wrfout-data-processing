```python
import pandas as pd
import numpy as np
from scipy.spatial.distance import cdist

# 读取 Excel 文件
df = pd.read_excel("E:\\000_FuJian\\rainnc\\A2.xlsx")

# 提取预测数据的经度、纬度和降水量
predict_data = df.iloc[:, [0, 1, 2]].values

# 提取观测数据的经度和纬度
observe_data = df.iloc[:, [4, 5]].values

# 计算观测数据和预测数据之间的距离
distances = cdist(observe_data, predict_data[:, :2], metric='euclidean')

# 找出每个观测数据对应的最近的预测数据索引
nearest_indexes = np.argmin(distances, axis=1)

# 获取最近的预测数据的降水量
nearest_rainfall = predict_data[nearest_indexes, 2]

# 插入到第7列
df['Predicted Rainfall'] = nearest_rainfall

# 将处理后的数据输出到 Excel 文件
df.to_excel("E:\\000_FuJian\\rainnc\\A2-1.xlsx", index=False)
```

```python
import os
import pandas as pd
import numpy as np
from scipy.spatial.distance import cdist

# 文件夹路径
folder_path = "E:\\000_FuJian\\000_21h\\rainnc-\\"

# 遍历文件夹中的所有文件
for filename in os.listdir(folder_path):
    if filename.endswith(".xlsx"):
        # 读取 Excel 文件
        df = pd.read_excel(os.path.join(folder_path, filename))

        # 提取预测数据的经度、纬度和降水量
        predict_data = df.iloc[:, [0, 1, 2]].values

        # 提取观测数据的经度和纬度
        observe_data = df.iloc[:, [4, 5]].values

        # 计算观测数据和预测数据之间的距离
        distances = cdist(observe_data, predict_data[:, :2], metric='euclidean')

        # 找出每个观测数据对应的最近的预测数据索引
        nearest_indexes = np.argmin(distances, axis=1)

        # 获取最近的预测数据的降水量
        nearest_rainfall = predict_data[nearest_indexes, 2]

        # 插入到第7列
        df['Predicted Rainfall'] = nearest_rainfall

        # 将处理后的数据输出到 Excel 文件
        output_filename = os.path.join(folder_path, filename.replace(".xlsx", "1.xlsx"))
        df.to_excel(output_filename, index=False)
        print('打饭(　o=^•ェ•)o　┏━┓')
print('干饭(　o=^•ェ•)o　┏━┓')
```
