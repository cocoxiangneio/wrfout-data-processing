```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point

# 加载Excel文件
file_path = r"E:/地信11111/作业/公交线路可视化/中山经纬/合并表格.xls"
excel_data = pd.ExcelFile(file_path, engine='xlrd')

# 初始化一个空的GeoDataFrame列表
gdf_list = []

# 遍历所有表格
for sheet in excel_data.sheet_names:
    # 读取当前表格
    df = pd.read_excel(file_path, sheet_name=sheet, engine='xlrd')
    
    # 提取经纬度数据，假设经度在D列（第3列），纬度在E列（第4列）
    df = df.rename(columns={df.columns[3]: 'Longitude', df.columns[4]: 'Latitude'})
    
    # 过滤掉没有经纬度的数据
    df = df.dropna(subset=['Longitude', 'Latitude'])
    
    # 创建几何点
    geometry = [Point(xy) for xy in zip(df['Longitude'], df['Latitude'])]
    
    # 创建GeoDataFrame
    geo_df = gpd.GeoDataFrame(df, geometry=geometry)
    
    # 将当前GeoDataFrame追加到列表中
    gdf_list.append(geo_df)

# 合并所有GeoDataFrame
gdf = pd.concat(gdf_list, ignore_index=True)

# 设置坐标系，例如WGS84
gdf.set_crs(epsg=4326, inplace=True)

# 保存为Shapefile
output_path = r"E:/大三上/城乡规划/城乡test/1.shp"
gdf.to_file(output_path)

output_path
```

```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
import os

# 加载Excel文件
file_path = r"E:/地信11111/作业/公交线路可视化/中山经纬/合并表格.xls"
excel_data = pd.ExcelFile(file_path, engine='xlrd')

# 初始化一个空的GeoDataFrame列表
gdf_list = []

# 遍历所有表格
for sheet in excel_data.sheet_names:
    # 读取当前表格
    df = pd.read_excel(file_path, sheet_name=sheet, engine='xlrd')
    
    # 提取经纬度数据，假设经度在D列（第3列），纬度在E列（第4列）
    df = df.rename(columns={df.columns[3]: 'Longitude', df.columns[4]: 'Latitude'})
    
    # 过滤掉没有经纬度的数据
    df = df.dropna(subset=['Longitude', 'Latitude'])
    
    # 创建几何点
    geometry = [Point(xy) for xy in zip(df['Longitude'], df['Latitude'])]
    
    # 创建GeoDataFrame
    geo_df = gpd.GeoDataFrame(df, geometry=geometry)
    
    # 重命名列，确保列名是ASCII字符
    geo_df.columns = [col.encode('ascii', 'ignore').decode('ascii') for col in geo_df.columns]
    
    # 将当前GeoDataFrame追加到列表中
    gdf_list.append(geo_df)

# 合并所有GeoDataFrame
gdf = pd.concat(gdf_list, ignore_index=True)

# 设置坐标系，例如WGS84
gdf.set_crs(epsg=4326, inplace=True)

# 定义输出路径
output_path = r"E:/大三上/城乡规划/城乡test/10.shp"

# 检查并创建目录（如果不存在）
output_dir = os.path.dirname(output_path)
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

# 保存为Shapefile
gdf.to_file(output_path)

output_path
```
