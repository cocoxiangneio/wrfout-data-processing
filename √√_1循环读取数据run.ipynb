```python
import os
import netCDF4 as nc
import pandas as pd
from wrf import getvar

# 指定包含文件的文件夹路径
folder_path = "E:/000_FuJian/wrfout_FuJian/d02"

# 遍历文件夹中的所有文件
for file_name in os.listdir(folder_path):
    if file_name.endswith('.nc'):
        # 读取 NetCDF 文件
        file_path = os.path.join(folder_path, file_name)
        ncfile = nc.Dataset(file_path)

        # 处理数据
        lats = getvar(ncfile, "XLAT",timeidx=0)[0:199,24:165]
        lons = getvar(ncfile, "XLONG",timeidx=0)[0:199,24:165]
        rain1 = getvar(ncfile, "RAINNC",21)[0:199,24:165]  +  getvar(ncfile, "RAINC",21)[0:199,24:165]
        rain2 = getvar(ncfile, "RAINNC",0)[0:199,24:165]  +  getvar(ncfile, "RAINC",0)[0:199,24:165]
        rain = rain1 - rain2

        lats_arr = lats.values.flatten()
        lons_arr = lons.values.flatten()
        rain_arr = rain.values.flatten()

        # 将数据保存为 Excel 文件
        output_file_name = file_name.replace('.nc', '.xlsx')
        output_file_path = os.path.join("E:/000_FuJian/000_21h", output_file_name)
        
        df = pd.DataFrame({'Longitude': lons_arr, 'Latitude': lats_arr, 'Rainfall': rain_arr})
        df.to_excel(output_file_path, index=False)

        print(f'文件 {file_name} 处理完成，并保存为 {output_file_name}')
```
