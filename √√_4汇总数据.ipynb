```python
import os
import openpyxl

# 创建一个字典来存储每个 Excel 文件的第7列数据，键为文件名
data_dict = {}

# 待读取数据的文件夹路径
folder_path = r"E:\000_FuJian\000_21h\rainnc-1"

# 获取文件夹内所有的 Excel 文件
for file_name in os.listdir(folder_path):
    if file_name.endswith(".xlsx") or file_name.endswith(".xls"):
        file_path = os.path.join(folder_path, file_name)
        wb = openpyxl.load_workbook(file_path)
        ws = wb.active
        data_list = []
        for row in ws.iter_rows(min_row=2, min_col=7, max_col=7, values_only=True):
            data_list.append(row[0])
        data_dict[file_name] = data_list

# 创建新的 Excel 文件
new_wb = openpyxl.Workbook()
new_ws = new_wb.active

# 将数据写入新的 Excel 文件，从第三列开始写入
for col_idx, (file_name, data_list) in enumerate(data_dict.items()):
    new_ws.cell(row=1, column=col_idx+4, value=file_name)  # 将文件名作为第一行，从第三列开始写入
    for row_idx, value in enumerate(data_list):
        new_ws.cell(row=row_idx+1, column=col_idx+4, value=value)  # 写入数据，从第三列开始写入
        
# 将数据写入新的 Excel 文件
for col_idx, (file_name, data_list) in enumerate(data_dict.items()):
    new_ws.cell(row=1, column=col_idx+4, value=file_name)  # 将文件名作为第一行
    for row_idx, value in enumerate(data_list):
        new_ws.cell(row=row_idx+2, column=col_idx+4, value=value)  # 写入数据


# 保存新的 Excel 文件
new_wb.save('E:\\000_FuJian\\000_21h\\rainnc-1\\21汇总.xlsx')

print("所有 Excel 文件的第7列数据已成功合并并写入新的 Excel 文件 'combined_data.xlsx' 中，并从第三列开始写入！")

```
