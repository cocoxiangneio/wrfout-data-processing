```python
import pandas as pd

# 定义阈值变量
threshold_value = 0.1

file_path = 'E:/000_FuJian/00_汇总_run/03汇总.xlsx'

df = pd.read_excel(file_path)
column_names = df.columns

# 创建用于多个工作表
output_path ='E:/000_FuJian/00_汇总_run/03NA.xlsx'
with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
    # 使用循环来动态设置筛选条件
    for i in range(3, 30):
        if i < len(column_names):  # 确保列索引在范围内
            filtered_df = df[(df[column_names[2]] > threshold_value) & (df[column_names[i]] > threshold_value)]
            selected_columns = list(column_names[:3]) + [column_names[i]]
            filtered_df = filtered_df[selected_columns]
            sheet_name = f"Sheet_{i}"
            filtered_df.to_excel(writer, sheet_name=sheet_name, index=False)

print(f"所有筛选后的数据已保存到 {output_path}")

# NB
output_path ='E:/000_FuJian/00_汇总_run/03NB.xlsx'
with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
    # 使用循环来动态设置筛选条件
    for i in range(3, 30):
        if i < len(column_names):  # 确保列索引在范围内
            filtered_df = df[(df[column_names[2]] < threshold_value) & (df[column_names[i]] > threshold_value)]
            selected_columns = list(column_names[:3]) + [column_names[i]]
            filtered_df = filtered_df[selected_columns]
            sheet_name = f"Sheet_{i}"
            filtered_df.to_excel(writer, sheet_name=sheet_name, index=False)
            
print(f"所有筛选后的数据已保存到 {output_path}")

# NC
output_path ='E:/000_FuJian/00_汇总_run/03NC.xlsx'
with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
    # 使用循环来动态设置筛选条件
    for i in range(3, 30):
        if i < len(column_names):  # 确保列索引在范围内
            filtered_df = df[(df[column_names[2]] > threshold_value) & (df[column_names[i]] < threshold_value)]
            selected_columns = list(column_names[:3]) + [column_names[i]]
            filtered_df = filtered_df[selected_columns]
            sheet_name = f"Sheet_{i}"
            filtered_df.to_excel(writer, sheet_name=sheet_name, index=False)

print(f"所有筛选后的数据已保存到 {output_path}")
```

```python
#单个时间段，多个降雨阈值

import pandas as pd
import os

# 定义阈值变量
threshold_values = [0.1, 10, 25, 50, 100]

# 定义文件夹路径
folder_path = 'E:/000_FuJian/00_汇总_run'
TS_path = 'E:/000_FuJian/00_检验_run'
# 定义文件名前缀变量
file_prefix = '24'
file_path = os.path.join(folder_path, f"{file_prefix}汇总.xlsx")

df = pd.read_excel(file_path)
column_names = df.columns

# 创建用于多个工作表
for threshold_value in threshold_values:
    for prefix in ['NA', 'NB', 'NC']:
        output_path = os.path.join(TS_path, f"{file_prefix}{prefix}_{threshold_value}.xlsx")
        with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
            # 使用循环来动态设置筛选条件
            for i in range(3, 30):
                if i < len(column_names):  # 确保列索引在范围内
                    if prefix == 'NA':
                        filtered_df = df[(df[column_names[2]] > threshold_value) & (df[column_names[i]] > threshold_value)]
                    elif prefix == 'NB':
                        filtered_df = df[(df[column_names[2]] < threshold_value) & (df[column_names[i]] > threshold_value)]
                    elif prefix == 'NC':
                        filtered_df = df[(df[column_names[2]] > threshold_value) & (df[column_names[i]] < threshold_value)]
                        
                    selected_columns = list(column_names[:3]) + [column_names[i]]
                    filtered_df = filtered_df[selected_columns]
                    sheet_name = f"Sheet_{i}"
                    filtered_df.to_excel(writer, sheet_name=sheet_name, index=False)

        print(f"所有筛选后的数据已保存到 {output_path}")
        
print('done')
```
