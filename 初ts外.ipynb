```python
import netCDF4 as nc
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

file = r"E:/000_FuJian/wrfout_FuJian/C6_mp6_cu14/wrfout_d02.nc"
dataset = nc.Dataset(file)
all_vars = dataset.variables.keys()
#print(len(all_vars))
#print(all_vars)

# 获取所有变量信息
all_vars_info = dataset.variables.items()
all_vars_info = list(all_vars_info)
#print(all_vars_info)

# 获取时间信息
times = dataset.variables['Times'][:]
print(len(times))
print(times[0])
print(type(times[0]))
print(times[-1])
#2023-09-07_06:00:00
#2023-09-08_06:00:00


# 获取单个数据的值
# 总降水=RAINNC + RAINC + RAINSH
rainc = dataset.variables['RAINC'][:]
rainnc = dataset.variables['RAINNC'][:]
rainsh = dataset.variables['RAINSH'][:]
print('')


rainc_total = []
for i, value_outside in enumerate(rainc):
    for j, value_inside in enumerate(value_outside):
        for k, value in enumerate(value_inside):
            rainc_total.append(value)

rainnc_total = []
for i, value_outside in enumerate(rainnc):
    for j, value_inside in enumerate(value_outside):
        for k, value in enumerate(value_inside):
            rainnc_total.append(value)

rainsh_total = []
for i, value_outside in enumerate(rainsh):
    for j, value_inside in enumerate(value_outside):
        for k, value in enumerate(value_inside):
            rainsh_total.append(value)


# 计算总降水
rainfall_predicted = np.array(rainc_total) + np.array(rainnc_total) + np.array(rainsh_total)
print(len(rainfall_predicted))
#print(rainfall_predicted)


x = range(0, len(rainfall_predicted))
plt.plot(x, rainfall_predicted)
plt.show()
```
