```python
###合并NANBNC
import pandas as pd
import os

# 文件夹路径
directory = r"E:\000_FuJian\00_检验整理_run"

# 方案列表
schemes = ["a1", "a2", "b1", "b2", "b3", "b4", "b5", "b6", "b7", "b8", "b9", "b10", "b11", "b12", "c1", "c2", "c3", "c4", "c5", "c6"]

# 对每个方案的文件进行处理
for scheme in schemes:
    # 构建方案文件的路径
    file_paths = [
        os.path.join(directory, f"汇总NA_{scheme}.xlsx"),
        os.path.join(directory, f"汇总NB_{scheme}.xlsx"),
        os.path.join(directory, f"汇总NC_{scheme}.xlsx")
    ]

    # 读取所有Excel文件并合并成一个DataFrame
    dfs = [pd.read_excel(file) for file in file_paths if os.path.exists(file)]
    if dfs:
        combined_df = pd.concat(dfs)

        # 保存合并后的DataFrame到一个新的Excel文件
        output_path = os.path.join(directory, f"汇总NANBNC_{scheme}.xlsx")
        combined_df.to_excel(output_path, index=False)
        print(f"方案 {scheme} 合并完成，结果保存在: {output_path}")
    else:
        print(f"方案 {scheme} 不存在对应的文件，未进行合并")
```
